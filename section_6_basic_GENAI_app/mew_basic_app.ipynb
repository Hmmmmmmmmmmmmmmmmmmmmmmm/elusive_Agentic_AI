# Basic GEN-AI app

basic 4 steps first:
> Ingest > split > embed > Store

```python
#imports
from langchain_community.document_loaders import WebBaseLoader
```

```python
loader = WebBaseLoader("https://en.wikipedia.org/wiki/Cat")
docs = loader.load()
```

```python
docs
```

```python
print("  metadata:", docs[0].metadata)
print("  content chars:", len(docs[0].page_content))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 300, chunk_overlap=40)
```

```python
chunks = text_splitter.split_documents(docs)
```

```python
chunks
```

```python
print("Step 2 - chunks:", len(chunks))
print("  chunk sizes:", [len(c.page_content) for c in chunks], "...")
```

```python
from langchain_ollama.embeddings import OllamaEmbeddings
```

```python
emb = OllamaEmbeddings(model='nomic-embed-text',num_gpu=-1)
```

```python
vec = emb.embed_documents([c.page_content for c in chunks])
print("Step 3 - vectors:", len(vec), "| dims each:", len(vec[0]))
```

```python
from langchain_community.vectorstores import FAISS
import shutil
```

```python
def reset_index(folder):
    shutil.rmtree(folder, ignore_errors=True)
    print(f"  (cleared old {folder}/ if present)")
```

```python
vs_faiss = FAISS.from_documents(chunks, emb)
reset_index("vector_index_cats_wiki")
vs_faiss.save_local("vector_index_cats_wiki")
print("FAISS index size:", vs_faiss.index.ntotal, "-> saved to disk")
```

```python
query = "What is a cat?"
```

```python
results = vs_faiss.similarity_search_with_score(query=query)
results
```

```python
# llm:
from langchain_ollama.chat_models import ChatOllama
llm = ChatOllama(model="mistral:7b ")
```

```python
# Retrieval Chain

from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are a cat expert who answer based on: {context}"), #system
        ("human","{input}"), #human
    ]
)

documents_chain = create_stuff_documents_chain(
    prompt=prompt,
    llm=llm
)

documents_chain
```

```python
# retriever

retriever = vs_faiss.as_retriever()
```

```python
from langchain_classic.chains import create_retrieval_chain
retriever_chain = create_retrieval_chain(retriever,documents_chain)
```

```python
retriever_chain
```

```python
response = retriever_chain.invoke(
    {"input":"what does a cat say?"}
)
```
