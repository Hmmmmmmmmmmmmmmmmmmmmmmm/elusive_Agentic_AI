Where each connects to what you already learned
- Retriever (1) is literally step 4 (vectorstore) wearing a uniform invoke() interface. Nothing new to learn — just a rename + one method.
- Prompt (2) is the first thing you've not seen yet. It's pure string templating.
- Chat LLM (3) is the one concept that might trip you up: you learned embeddings return vectors; chat models read and write text. They're different model families.
- LCEL (4) is the biggest conceptual leap — the | composition and Runnables. This is where "chain" really clicks.
- Provenance (5) we already spotted a real bug (duplicate chunks) — we'll fix it here.
- Memory (6) and agents (7) are where RAG graduates into the Agentic half.
